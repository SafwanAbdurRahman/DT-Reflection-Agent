```python
def start_reflection():
    import json
    try:
        with open('reflection-tree.json', 'r') as f:
            data = json.load(f)
    except FileNotFoundError:
        print("ERROR: Run the FIRST cell again to create the JSON file!")
        return

    nodes = {n['id']: n for n in data['nodes']}
    current_id = "START"

    print("--- AGENT ACTIVATED ---")

    while True:
        node = nodes.get(current_id)
        if not node:
            print(f"Error: Node {current_id} not found in JSON.")
            break

        print(f"\n>> {node['text']}")

        if node['type'] == 'end':
            break

        if node['type'] == 'question':
            for i, opt in enumerate(node['options']):
                print(f"   {i+1}) {opt['text']}")

            # This logic forces the input box to appear clearly
            user_input = input("TYPE 1 OR 2 AND HIT ENTER: ")

            try:
                idx = int(user_input) - 1
                current_id = node['options'][idx]['target']
            except (ValueError, IndexError):
                print("Invalid input. Try again.")
                continue
        else:
            current_id = node.get('next') or node.get('target')

start_reflection()
```
